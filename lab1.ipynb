# Lab-4: Flutter Widgets Theory

## **MaterialApp & Scaffold**

- **MaterialApp**  
  The root of the Flutter app that provides **material design elements**.

- **Scaffold**  
  Provides a **basic structure** for the app including `AppBar`, `Body`, `Drawer`, etc.

---

## **AppBar**

The `AppBar` widget creates a **material design app bar** at the top of the screen.

**Properties used in this lab:**

- `centerTitle: true` → Centers the title.
- `title` → Displays the text **“Lab-4”** with custom styling (`TextStyle`).

---

## **Container**

`Container` is a versatile widget used for **layout, decoration, and spacing**.

**Common properties:**

- `height` and `width` → Defines the size of the container.
- `color` → Sets background color.
- `margin` → Adds space **outside** the container.
- `padding` → Adds space **inside** the container.

---

## **Column**

`Column` is a widget that arranges its children **vertically**.

**Properties:**

- `mainAxisAlignment` → Controls **vertical alignment** of children (`spaceAround` is used here to distribute space evenly).  
- `crossAxisAlignment` → Controls **horizontal alignment** (`center` in this lab).

---

## **Expanded**

`Expanded` makes a child of `Column` or `Row` **fill the available space**.  

- In this lab, it is used to make the **middle container grow** and occupy the remaining space between the top and bottom containers.

---

## **Center**

`Center` widget **centers its child** both vertically and horizontally within its parent container.

---

## **Widgets Used in the Lab**

- **Text** → Displays text inside a widget.  
- **EdgeInsets** → Provides **padding and margin** values.  
- **Color.fromARGB** → Custom color using **alpha, red, green, blue** values.
